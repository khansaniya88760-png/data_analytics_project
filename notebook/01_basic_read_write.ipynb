## 1. Reading and writing a plain text file (pure Python)

```python
# 'w' = write (creates/overwrites the file)
with open("notes.txt", "w") as f:
    f.write("Online Shopping Data Analysis\n")
    f.write("Dataset: shopping_data.csv\n")

# 'a' = append (adds to the end)
with open("notes.txt", "a") as f:
    f.write("Step 1: Learn read/write operations\n")

# 'r' = read
with open("notes.txt", "r") as f:
    print(f.read())
```

## 2. Reading the CSV file with pandas

```python
import pandas as pd
```

```python
df = pd.read_csv("shopping_data.csv")
df.head()          # first 5 rows
```

```python
df.tail(3)       # last 3 rows
```

```python
print('Shape (rows, columns):', df.shape)
```

```python
print('Columns:', list(df.columns))
```

```python
df.info()        # column names, data types, non-null counts
```

```python
df.describe()    # summary statistics of numeric columns
```

## 3. Writing data to files

```python
# a) Save only some columns to a new CSV
small = df[["Order_ID", "City", "Product_Category", "Order_Value_INR"]]
small.to_csv("orders_small.csv", index=False)   # index=False -> don't write row numbers
print("orders_small.csv saved")
```

```python
%pip install openpyxl
```

```python
# b) Save the first 20 rows as Excel (needs: pip install openpyxl)
df.head(20).to_excel("orders_sample.xlsx", index=False)
print("orders_sample.xlsx saved")
```

```python
# c) Save as JSON
df.head(20).to_json("orders_sample.json", orient="records", indent=2)
print("orders_sample.json saved")
```

## 4. Reading the saved files back

```python
pd.read_csv('orders_small.csv').head()
```

```python
pd.read_excel('orders_sample.xlsx').head()
```

```python
pd.read_json('orders_sample.json').head()
```

## 5. Useful options of `read_csv`

```python
# Read only some columns and only 10 rows
pd.read_csv("shopping_data.csv", usecols=["Order_ID", "City", "Order_Value_INR"], nrows=10)
```
