### 2. Filtering rows

```python
# Orders above Rs. 10,000
df[df["Order_Value_INR"] > 10000].head()
```

```python
# Electronics orders paid by UPI  (use & for AND, | for OR)
df[(df["Product_Category"] == "Electronics") & (df["Payment_Mode"] == "UPI")].head()
```

### 3. Exploring categories

```python
df["Product_Category"].value_counts()
```

```python
df["Payment_Mode"].value_counts(normalize=True) * 100   # percentages
```

```python
df["City"].unique()    # notice: same city written in different ways!
```

### 4. Finding data problems

```python
df.isnull().sum()        # missing values per column
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Impossible discount values
df[(df["Discount_Percent"] < 0) | (df["Discount_Percent"] > 100)]
```

### 5. Sorting

```python
df.sort_values("Order_Value_INR", ascending=False).head(5)
```

### 6. GroupBy (very important for analysis)

```python
df.groupby("Product_Category")["Order_Value_INR"].mean().round(2)
```

```python
df.groupby("Payment_Mode")["Customer_Rating"].mean().round(2)
```

```python
df.groupby("Product_Category").agg(
    total_orders=("Order_ID", "count"),
    avg_value=("Order_Value_INR", "mean"),
    avg_rating=("Customer_Rating", "mean")
).round(2)
```

### 7. Creating a new column

```python
df["Is_Big_Order"] = df["Order_Value_INR"] > 5000
df[["Order_ID", "Order_Value_INR", "Is_Big_Order"]].head()
```

## Part B: NumPy

```python
values = df["Order_Value_INR"].dropna().to_numpy()   # pandas column -> NumPy array
print(type(values), values.shape)
```

```python
print("Mean   :", np.mean(values))
print("Median :", np.median(values))
print("Std dev:", np.std(values))
print("Min/Max:", np.min(values), np.max(values))
```

```python
# Percentiles
print("25th, 50th, 75th percentile:", np.percentile(values, [25, 50, 75]))
```

```python
# np.where = if/else for every element
df["Order_Size"] = np.where(df["Order_Value_INR"] > 5000, "High", "Normal")
df["Order_Size"].value_counts()
```

```python
# Replace impossible discounts with NaN using NumPy
df["Discount_Clean"] = np.where(
    (df["Discount_Percent"] < 0) | (df["Discount_Percent"] > 100),
    np.nan,
    df["Discount_Percent"]
)
df["Discount_Clean"].describe()
```

```python
# Correlation between numeric columns
num_cols = ["Items_Count", "Order_Value_INR", "Discount_Percent", "Delivery_Days", "Browsing_Time_Min", "Customer_Rating"]
df[num_cols].corr().round(2)
```

```python
# Basic NumPy array maths
a = np.array([10, 20, 30, 40])
print(a * 2, a + 5, a.sum(), a.mean())
```
